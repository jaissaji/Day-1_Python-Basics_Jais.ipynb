## 1. Variables and arithmetic

## 2. Data types and comparisons

```python
a = 23
b = 23.5
c = "23.5"
print(type(a), type(b), type(c))
print(34 > 45)
print(34 != 45)
```

## 3. Lists, tuples and dictionaries

```python
my_list = ["Vasiullah", "Ritesh", "Sanjana"]
my_list.append("Tessica")
print(my_list)
print(my_list[1], my_list[-1])

my_tuple = ("Ram", "Sham", [2, 3, 4, 5], 25, 25.5)
my_tuple[2][1] = 20
print(my_tuple)

student = {"name": "Jais", "course": "Machine Learning", "marks": 85}
print(student)
```

## 4. Loops and list comprehension

```python
planets = [["Mercury", "Venus", "Earth"], ["Mars", "Jupiter", "Saturn"], ["Uranus", "Neptune", "Pluto"]]

short_planets = []
for sublist in planets:
    for planet in sublist:
        if len(planet) < 6:
            short_planets.append(planet)
print(short_planets)

long_planets = [planet for sublist in planets for planet in sublist if len(planet) > 6]
print(long_planets)
```

## 5. Functions

```python
def square(n):
    return n ** 2

for n in [1, 2, 3, 4, 5]:
    print(n, square(n))
```
