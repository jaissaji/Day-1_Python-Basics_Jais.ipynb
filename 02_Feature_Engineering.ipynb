# Feature Engineering – Titanic

Create meaningful features from existing Titanic variables and explain why each engineered feature may help a model.

```python
import pandas as pd
import numpy as np
import seaborn as sns

df = sns.load_dataset("titanic")
df.head()
```

## 1. Create new features

```python
df_fe = df.copy()

# Family size = siblings/spouse + parents/children + passenger
df_fe["family_size"] = df_fe["sibsp"] + df_fe["parch"] + 1

# Whether the passenger travelled alone
df_fe["is_alone"] = (df_fe["family_size"] == 1).astype(int)

# Age bands
df_fe["age_group"] = pd.cut(
    df_fe["age"],
    bins=[0, 12, 18, 35, 60, np.inf],
    labels=["Child", "Teen", "Young Adult", "Adult", "Senior"]
)

# Fare per family member
df_fe["fare_per_person"] = df_fe["fare"] / df_fe["family_size"].replace(0, np.nan)

df_fe[["survived", "family_size", "is_alone", "age_group", "fare_per_person"]].head()
```

## 2. Check whether the engineered features are useful

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
sns.barplot(data=df_fe, x="is_alone", y="survived", ax=ax[0])
ax[0].set_title("Survival rate by travelling alone")

sns.barplot(data=df_fe, x="family_size", y="survived", ax=ax[1])
ax[1].set_title("Survival rate by family size")
plt.tight_layout()
plt.show()
```

### Explanation
- **family_size** captures household/group structure.
- **is_alone** summarizes whether a passenger travelled independently.
- **age_group** converts a continuous age variable into interpretable life-stage categories.
- **fare_per_person** adjusts fare for group size.
