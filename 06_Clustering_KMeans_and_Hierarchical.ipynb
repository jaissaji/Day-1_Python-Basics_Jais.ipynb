# Unsupervised Learning – K-Means and Hierarchical Clustering

Perform customer-style segmentation using standardized numeric features, then compare K-Means and agglomerative clustering.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score

X, _ = make_blobs(n_samples=300, centers=4, cluster_std=1.2, random_state=42)
X = StandardScaler().fit_transform(X)
```

## K-Means

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
labels_k = kmeans.fit_predict(X)
print("K-Means silhouette score:", silhouette_score(X, labels_k))

plt.scatter(X[:,0], X[:,1], c=labels_k)
plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1], marker="X", s=150)
plt.title("K-Means clusters")
plt.show()
```

## Hierarchical clustering

```python
agg = AgglomerativeClustering(n_clusters=4, linkage="ward")
labels_h = agg.fit_predict(X)
print("Hierarchical silhouette score:", silhouette_score(X, labels_h))

plt.scatter(X[:,0], X[:,1], c=labels_h)
plt.title("Agglomerative clusters")
plt.show()
```

### Key ideas
K-Means partitions data around centroids. Hierarchical clustering builds nested groups by repeatedly merging similar observations/clusters. Scaling matters because distance drives both methods.
