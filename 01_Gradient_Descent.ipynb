# Gradient Descent from Scratch

Implement gradient descent for a simple linear regression objective and visualize convergence.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
X = np.linspace(0, 10, 100)
y = 3 * X + 5 + rng.normal(0, 1, size=len(X))

w, b = 0.0, 0.0
learning_rate = 0.01
history = []

for epoch in range(1000):
    y_hat = w * X + b
    error = y_hat - y
    loss = np.mean(error ** 2)
    dw = 2 * np.mean(error * X)
    db = 2 * np.mean(error)
    w -= learning_rate * dw
    b -= learning_rate * db
    history.append(loss)

print("Learned slope:", w)
print("Learned intercept:", b)
```

```python
plt.plot(history)
plt.xlabel("Epoch")
plt.ylabel("MSE")
plt.title("Gradient descent convergence")
plt.show()

plt.scatter(X, y, s=15)
plt.plot(X, w * X + b)
plt.xlabel("X")
plt.ylabel("y")
plt.title("Fitted regression line")
plt.show()
```

### Key idea
Gradient descent updates parameters in the direction opposite to the gradient of the loss. The learning rate controls the step size.
