# Feed-Forward Neural Network and Backpropagation Demo

A compact conceptual demonstration of forward propagation, loss and gradient-based weight updates using a small XOR problem.

```python
import numpy as np

# XOR dataset
X = np.array([[0,0], [0,1], [1,0], [1,1]], dtype=float)
y = np.array([[0], [1], [1], [0]], dtype=float)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_derivative(a):
    return a * (1 - a)

rng = np.random.default_rng(1)
W1 = rng.normal(0, 0.5, (2, 4))
b1 = np.zeros((1, 4))
W2 = rng.normal(0, 0.5, (4, 1))
b2 = np.zeros((1, 1))

lr = 0.5
losses = []

for epoch in range(10000):
    # Forward propagation
    z1 = X @ W1 + b1
    a1 = sigmoid(z1)
    z2 = a1 @ W2 + b2
    a2 = sigmoid(z2)

    # Mean squared error
    loss = np.mean((a2 - y) ** 2)
    losses.append(loss)

    # Backpropagation
    d2 = (a2 - y) * sigmoid_derivative(a2)
    dW2 = a1.T @ d2 / len(X)
    db2 = np.mean(d2, axis=0, keepdims=True)

    d1 = (d2 @ W2.T) * sigmoid_derivative(a1)
    dW1 = X.T @ d1 / len(X)
    db1 = np.mean(d1, axis=0, keepdims=True)

    W2 -= lr * dW2
    b2 -= lr * db2
    W1 -= lr * dW1
    b1 -= lr * db1

pred_prob = sigmoid(sigmoid(X @ W1 + b1) @ W2 + b2)
pred = (pred_prob >= 0.5).astype(int)

print("Probabilities:\n", np.round(pred_prob, 3))
print("Predictions:", pred.ravel())
print("Actual:", y.ravel())
```

```python
import matplotlib.pyplot as plt
plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("MSE loss")
plt.title("Backpropagation convergence")
plt.show()
```

### Key idea
Forward propagation computes predictions layer by layer. Backpropagation computes how the loss changes with respect to each weight, and gradient descent updates the weights in the direction that reduces the loss.
