# Missing Data Analysis and Imputation

Identify missing values, visualize them, and compare simple imputation strategies. The example uses the Titanic dataset so the notebook is reproducible without uploading a CSV.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("titanic")
df.head()
```

## 1. Inspect missing values

```python
missing = df.isna().sum().sort_values(ascending=False)
missing[missing > 0]
```

```python
plt.figure(figsize=(10,4))
sns.heatmap(df.isna(), cbar=False)
plt.title("Missing-value pattern")
plt.show()
```

## 2. Numeric and categorical imputation

```python
from sklearn.impute import SimpleImputer

numeric_cols = df.select_dtypes(include=np.number).columns
categorical_cols = df.select_dtypes(exclude=np.number).columns

df_imputed = df.copy()
df_imputed[numeric_cols] = SimpleImputer(strategy="median").fit_transform(df_imputed[numeric_cols])
df_imputed[categorical_cols] = SimpleImputer(strategy="most_frequent").fit_transform(df_imputed[categorical_cols])

print("Missing values after imputation:", df_imputed.isna().sum().sum())
```

## 3. Compare before and after

```python
comparison = pd.DataFrame({
    "before": df.isna().sum(),
    "after": df_imputed.isna().sum()
})
comparison[comparison["before"] > 0]
```

### Conclusion
Median imputation is appropriate for numeric variables when outliers may exist; most-frequent imputation is a simple baseline for categorical variables. In a production model, imputation should be fitted on the training data only.
